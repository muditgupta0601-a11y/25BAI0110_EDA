```python
# FACTORIAL

num = int(input("Enter a number: "))

fact = 1
for i in range(1, num + 1):
    fact *= i

print("Factorial =", fact)
```

```python
# TABLE
num = int(input("Enter a number: "))

for i in range(1, 11):
    print(num, "x", i, "=", num * i)
```

```python
# PRIME NO.
num = int(input("Enter a number: "))

if num > 1:
    for i in range(2, num):
        if num % i == 0:
            print("Not Prime")
            break
    else:
        print("Prime")
else:
    print("Not Prime")
```

```python
# SUM OF N NO.
n = int(input("Enter a number: "))

sum = 0
for i in range(1, n + 1):
    sum += i

print("Sum =", sum)
```

```python
# CHECK EVEN ODD
num = int(input("Enter a number: "))

if num % 2 == 0:
    print("Even")
else:
    print("Odd")
```

```python
# PALINDROME
num = int(input("Enter a number: "))

temp = num
rev = 0

while num > 0:
    digit = num % 10
    rev = rev * 10 + digit
    num = num // 10

if temp == rev:
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
# ARMSTRONG
num = int(input("Enter a number: "))

temp = num
sum = 0

while temp > 0:
    digit = temp % 10
    sum += digit ** 3
    temp //= 10

if sum == num:
    print("Armstrong Number")
else:
    print("Not Armstrong")
```

```python
# DIGIT COUNT
num = int(input("Enter a number: "))

count = 0

while num > 0:
    count += 1
    num //= 10

print("Digits =", count)
```

```python
#SUM OF DIGITS
num = int(input("Enter a number: "))

sum = 0

while num > 0:
    digit = num % 10
    sum += digit
    num //= 10

print("Sum =", sum)
```

```python
# REVERSE A NO.
num = int(input("Enter a number: "))

rev = 0

while num > 0:
    digit = num % 10
    rev = rev * 10 + digit
    num //= 10

print("Reverse =", rev)
```
