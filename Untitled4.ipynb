```python
import pandas as pd
import numpy as np
```

```python
bins = [100, 110, 120, 130, 140]
height = [105, 112, 118, 123, 108, 115, 128, 135, 101, 119, 126, 131]
category = pd.cut(height, bins)
print(category)
print(pd.value_counts(category))
```

```python
bin_names = ['Category A', 'Category B', 'Category C', 'Category D']
print(pd.cut(height, bins, labels=bin_names))
```

```python
print(pd.cut(np.random.rand(50), 5, precision=2))
```

```python
randomNumbers = np.random.rand(3000)
category3 = pd.qcut(randomNumbers, 5)
print(category3)
```
