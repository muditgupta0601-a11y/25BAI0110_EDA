```python
import pandas as pd
import numpy as np

# Create Data 
data = [
    [120, 135, 150, 145, 160, 170],
    [98, 110, 125, 130, 140, 150],
    [180, 175, 190, 200, 210, 220],
    [145, 155, 165, 175, 185, 195],
    [200, 210, 220, 230, 240, 250],
    [160, 170, 180, 190, 200, 210],
    [115, 125, 135, 145, 155, 165],
    [175, 185, 195, 205, 215, 225],
    [140, 150, 160, 170, 180, 190],
    [130, 140, 150, 160, 170, 180]
]

# Create DataFrame
dfx = pd.DataFrame(
    data,
    index=[
        'apple', 'banana', 'kiwi', 'grapes', 'mango',
        'orange', 'papaya', 'guava', 'pineapple', 'watermelon'
    ],
    columns=[
        'store1', 'store2', 'store3',
        'store4', 'store5', 'store6'
    ]
)

print(dfx)
print("Original DataFrame")
print(dfx)
```

```python
# Add Missing Values
dfx['store4'] = np.nan

dfx.loc['watermelon'] = np.arange(15, 21)

dfx.loc['oranges'] = np.nan

dfx['store5'] = np.nan

dfx.loc['apple', 'store4'] = 20

print("\nDataFrame After Adding Missing Values")
print(dfx)
```

```python
 #Check Missing Values
 #isnull()
print(dfx.isnull())
 #notnull()
print(dfx.notnull())
# Number of Missing Values in Each Column
print(dfx.isnull().sum())
# Total Missing Values
print(dfx.isnull().sum().sum())
# Count of Non-Missing Values
print(dfx.count())
```

```python
# Dropping Missing Values
#nStore4 Values Not Null
print(dfx.store4[dfx.store4.notnull()])
#Drop NaN from Store4
print(dfx.store4.dropna())
#Drop Rows Having Any NaN
print(dfx.dropna())
#Drop Rows Where All Values Are NaN
print(dfx.dropna(how='all'))
#Drop Columns Where All Values Are NaN
print(dfx.dropna(how='all', axis=1))
#Drop Columns Having Less Than 5 Non-NaN Values
print(dfx.dropna(thresh=5, axis=1))
```

```python
#  Mathematical Operations with NaN
ar1 = np.array([100, 200, np.nan, 300])
ser1 = pd.Series(ar1)
#Numpy Mean
print(np.nanmean(ar1))
#Pandas Series Mean
print(ser1.mean())
ser2 = dfx.store4
#Sum of Store4
print(ser2.sum())
#Average of Store4
print(ser2.mean())
#Cumulative Sum
print(ser2.cumsum())
```

```python
 #  Fill Missing Values
filledDf = dfx.fillna(0)
#DataFrame After Filling NaN with 0
print(filledDf)
#Mean Before Filling
print(dfx.mean(numeric_only=True))
#Mean After Filling
print(filledDf.mean(numeric_only=True))
```

```python
#  Forward Fill
#Forward Fill (store4)
print(dfx.store4.ffill())
```

```python
#  Backward Fill

#Backward Fill (store4)
print(dfx.store4.bfill())
```

```python
import numpy as np
import pandas as pd
# Enable Iterative Imputer
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import SimpleImputer, IterativeImputer
df =df = pd.DataFrame({
    "EmployeeID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "Age": [22, 25, np.nan, 40, 35, np.nan, 28, 31, np.nan, 45],
    "Salary": [25000, np.nan, 45000, 60000, np.nan, 50000, 55000, np.nan, 48000, 70000],
    "Department": ["HR", "IT", np.nan, "Finance", "IT", "HR", "Marketing", np.nan, "Finance", "IT"]
})

print("Original DataFrame")
print(df)
```

```python
# 1. Mean Imputation
mean_imputer = SimpleImputer(strategy="mean")
df["Age_Mean"] = mean_imputer.fit_transform(df[["Age"]])
```

```python
# 2. Median Imputation
median_imputer = SimpleImputer(strategy="median")
df["Salary_Median"] = median_imputer.fit_transform(df[["Salary"]])
```

```python
# 3. Mode Imputation
mode_imputer = SimpleImputer(strategy="most_frequent")
df["Department_Mode"] = mode_imputer.fit_transform(df[["Department"]]).ravel()
```

```python
# 4. MICE (Iterative Imputation)
# -------------------------------------
num_cols = ["Age", "Salary"]
mice_imputer = IterativeImputer(
    max_iter=10,
    random_state=42)
df_mice = df[num_cols].copy()
df_mice[:] = mice_imputer.fit_transform(df_mice)
df["Age_MICE"] = df_mice["Age"]
df["Salary_MICE"] = df_mice["Salary"]
```

```python
print("\nDataFrame After Imputation")
print(df)
```
